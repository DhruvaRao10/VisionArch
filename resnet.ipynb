```python
import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
from torch.utils.data import DataLoader
import torchvision.transforms as transforms
import torchvision.datasets as datasets
import matplotlib.pyplot as plt
import numpy as np
from tqdm import tqdm
import time

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

class CNN(nn.Module):
    """
    - Conv layers 
    - Batch norm
    - ReLU activation functions
    - Max pooling layers
    - Dropout for regularization
    - Fully connected layers
    """
    
    def __init__(self, num_classes=10, dropout_prob=0.5):
        super(CNN, self).__init__()
        
        # First Conv Block
        # Input: 1x28x28 -> Output: 32x28x28 
        self.conv1 = nn.Conv2d(
            in_channels=1,    
            out_channels=32,   
            kernel_size=3,     
            stride=1,         
            padding=1           
        )
        # Batch norm params 
        self.bn1 = nn.BatchNorm2d(32)
        
        # Second Conv Block
        # Input: 32x14x14 -> Output: 64x14x14
        self.conv2 = nn.Conv2d(32, 64, kernel_size=3, stride=1, padding=1)
        self.bn2 = nn.BatchNorm2d(64)
        
        # Third Conv Block
        # Input: 64x7x7 -> Output: 128x7x7
        self.conv3 = nn.Conv2d(64, 128, kernel_size=3, stride=1, padding=1)
        self.bn3 = nn.BatchNorm2d(128)
        
        # Max pooling layer 
        self.pool = nn.MaxPool2d(kernel_size=2, stride=2)
        # Dropout layer 
        self.dropout = nn.Dropout(dropout_prob)
        
        # Calculate the size after convolutions and pooling
        # 28x28 -> 14x14 -> 7x7 -> 3x3 
        # pooling is done after conv1 and conv2
        self.fc1_input_size = 128 * 7 * 7 
        
        # Fully connected layers 
        self.fc1 = nn.Linear(self.fc1_input_size, 512)
        self.fc2 = nn.Linear(512, 256)
        self.fc3 = nn.Linear(256, num_classes)
        
        # Initialize weights using He initialization for ReLU
        self._initialize_weights()
    
    def _initialize_weights(self):
        """
        Initialize weights using He initialization for ReLU networks
        and biases to zero as discussed in theory
        """
        for m in self.modules():
            if isinstance(m, nn.Conv2d):
                # He initialization: W ~ Normal(0, sqrt(2/n_in))
                nn.init.kaiming_normal_(m.weight, mode='fan_in', nonlinearity='relu')
                if m.bias is not None:
                    # Initialize biases to zero
                    nn.init.constant_(m.bias, 0)
            elif isinstance(m, nn.Linear):
                # He initialization for linear layers
                nn.init.kaiming_normal_(m.weight, mode='fan_in', nonlinearity='relu')
                nn.init.constant_(m.bias, 0)
            elif isinstance(m, nn.BatchNorm2d):
                # Initialize gamma=1, beta=0 for batch normalization
                nn.init.constant_(m.weight, 1)
                nn.init.constant_(m.bias, 0)
    
    def forward(self, x):
        """
        Forward pass implementing the mathematical operations:
        - Convolution: Y = X * K + b
        - Batch normalization: y = gamma * (x_normalized) + beta
        - ReLU: f(x) = max(0, x)
        - Max pooling: P = max(window)
        """
        
        # First conv block: Conv -> BatchNorm -> ReLU -> MaxPool
        # Z1 = X * W1 + b1
        x = self.conv1(x)
        # Batch normalization with learnable gamma, beta
        x = self.bn1(x)
        # ReLU activation: f(x) = max(0, x)
        x = F.relu(x)
        # Max pooling: reduces 28x28 to 14x14
        x = self.pool(x)
        
        # Second conv block
        x = self.conv2(x)
        x = self.bn2(x)
        x = F.relu(x)
        x = self.pool(x)  # 14x14 to 7x7
        
        # Third convolutional block (no pooling)
        x = self.conv3(x)
        x = self.bn3(x)
        x = F.relu(x)
        
        # Flatten for fully connected layers
        # Reshape from (batch_size, 128, 7, 7) to (batch_size, 128*7*7)
        x = x.view(x.size(0), -1)
        
        # First fully connected layer with dropout
        x = self.fc1(x)
        x = F.relu(x)
        x = self.dropout(x)
        
        # Second fully connected layer with dropout
        x = self.fc2(x)
        x = F.relu(x)
        x = self.dropout(x)
        
        # Output layer (no activation, will use CrossEntropyLoss)
        x = self.fc3(x)
        
        return x


def load_data(batch_size=128):
    """
    Load and preprocess MNIST dataset with data augmentation
    """
    # Data preprocessing and augmentation
    train_transform = transforms.Compose([
        transforms.RandomRotation(10),  # Random rotation ±10 degrees
        transforms.RandomAffine(0, translate=(0.1, 0.1)),  # Random translation
        transforms.ToTensor(),  # Convert to tensor and normalize to [0,1]
        transforms.Normalize((0.1307,), (0.3081,))  # MNIST mean and std
    ])
    
    test_transform = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize((0.1307,), (0.3081,))
    ])
    
    # Load datasets
    train_dataset = datasets.MNIST(
        root='./data', 
        train=True, 
        download=True, 
        transform=train_transform
    )
    
    test_dataset = datasets.MNIST(
        root='./data', 
        train=False, 
        download=True, 
        transform=test_transform
    )
    
    # Create data loaders
    train_loader = DataLoader(
        train_dataset, 
        batch_size=batch_size, 
        shuffle=True,
        num_workers=2,
        pin_memory=True
    )
    
    test_loader = DataLoader(
        test_dataset, 
        batch_size=batch_size, 
        shuffle=False,
        num_workers=2,
        pin_memory=True
    )
    
    return train_loader, test_loader


class LearningRateScheduler:
    """
    Implements cosine annealing learning rate schedule:
    η_t = η_min + (1/2)(η_max - η_min)(1 + cos(πt/T))
    """
    def __init__(self, optimizer, max_lr, min_lr, total_epochs):
        self.optimizer = optimizer
        self.max_lr = max_lr
        self.min_lr = min_lr
        self.total_epochs = total_epochs
        
    def step(self, epoch):
        lr = self.min_lr + 0.5 * (self.max_lr - self.min_lr) * \
             (1 + np.cos(np.pi * epoch / self.total_epochs))
        
        for param_group in self.optimizer.param_groups:
            param_group['lr'] = lr
        
        return lr


def calculate_accuracy(outputs, labels):
    """Calculate classification accuracy"""
    _, predicted = torch.max(outputs.data, 1)
    total = labels.size(0)
    correct = (predicted == labels).sum().item()
    return 100 * correct / total


def train_model(model, train_loader, test_loader, num_epochs=20, initial_lr=0.001):
    """
    Complete training loop with all optimization components:
    - Adam optimizer with learnable momentum parameters (beta1, beta2)
    - Cross-entropy loss
    - Learning rate scheduling
    - Gradient clipping
    - L2 regularization through weight_decay
    """
    
    # Cross-entropy loss (includes softmax)
    criterion = nn.CrossEntropyLoss()
    
    # Adam optimizer with initialized hyperparameters
    optimizer = optim.Adam(
        model.parameters(),
        lr=initial_lr,
        betas=(0.9, 0.999),  # β₁, β₂ parameters from theory
        eps=1e-8,            # ε parameter for numerical stability
        weight_decay=1e-4    # L2 regularization coefficient λ
    )
    
    # Learning rate scheduler
    scheduler = LearningRateScheduler(
        optimizer, 
        max_lr=initial_lr, 
        min_lr=initial_lr/10, 
        total_epochs=num_epochs
    )
    
    # Gradient clipping threshold
    max_grad_norm = 1.0
    
    # Training history
    train_losses = []
    train_accuracies = []
    test_accuracies = []
    learning_rates = [] 
    
    print("Starting training...")
    print(f"Model has {sum(p.numel() for p in model.parameters())} parameters")
    
    for epoch in range(num_epochs):
        model.train()
        running_loss = 0.0
        correct_train = 0
        total_train = 0
        
        # Update learning rate using cosine annealing
        current_lr = scheduler.step(epoch)
        learning_rates.append(current_lr)
        
        # Training loop with progress bar
        train_pbar = tqdm(train_loader, desc=f'Epoch {epoch+1}/{num_epochs}', 
                  disable=False, leave=True, ncols=100)

        
        for batch_idx, (data, target) in enumerate(train_pbar):
            data, target = data.to(device), target.to(device)
            
            # Zero gradients
            optimizer.zero_grad()
            
            # Forward pass
            output = model(data)
            
            # Calculate loss (cross-entropy)
            loss = criterion(output, target)
            
            # Backward pass - compute gradients
            loss.backward()
            
            # Gradient clipping to prevent exploding gradients
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_grad_norm)
            
            # Update parameters using Adam optimizer
            optimizer.step()
            
            # Statistics
            running_loss += loss.item()
            correct_train += calculate_accuracy(output, target) * target.size(0) / 100
            total_train += target.size(0)
            
            # Update progress bar
            train_pbar.set_postfix({
                'Loss': f'{loss.item():.4f}',
                'Acc': f'{100*correct_train/total_train:.2f}%',
                'LR': f'{current_lr:.6f}'
            })
        
        # Calculate epoch statistics
        epoch_loss = running_loss / len(train_loader)
        epoch_train_acc = 100 * correct_train / total_train
        
        train_losses.append(epoch_loss)
        train_accuracies.append(epoch_train_acc)
        
        # Evaluation on test set
        test_acc = evaluate_model(model, test_loader)
        test_accuracies.append(test_acc)
        
        print(f'Epoch [{epoch+1}/{num_epochs}]')
        print(f'Train Loss: {epoch_loss:.4f}, Train Acc: {epoch_train_acc:.2f}%, Test Acc: {test_acc:.2f}%')
        print(f'Learning Rate: {current_lr:.6f}')
        print('-' * 60)
    
    return {
        'train_losses': train_losses,
        'train_accuracies': train_accuracies,
        'test_accuracies': test_accuracies,
        'learning_rates': learning_rates
    }


def evaluate_model(model, test_loader):
    """Evaluate model on test set"""
    model.eval()
    correct = 0
    total = 0
    
    with torch.no_grad():
        for data, target in test_loader:
            data, target = data.to(device), target.to(device)
            outputs = model(data)
            correct += calculate_accuracy(outputs, target) * target.size(0) / 100
            total += target.size(0)
    
    accuracy = 100 * correct / total
    return accuracy


def plot_training_history(history):
    """Plot training metrics"""
    fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(15, 10))
    
    epochs = range(1, len(history['train_losses']) + 1)
    
    # Training loss
    ax1.plot(epochs, history['train_losses'], 'b-', label='Training Loss')
    ax1.set_title('Training Loss')
    ax1.set_xlabel('Epoch')
    ax1.set_ylabel('Loss')
    ax1.legend()
    ax1.grid(True)
    
    # Training and test accuracy
    ax2.plot(epochs, history['train_accuracies'], 'b-', label='Training Accuracy')
    ax2.plot(epochs, history['test_accuracies'], 'r-', label='Test Accuracy')
    ax2.set_title('Training and Test Accuracy')
    ax2.set_xlabel('Epoch')
    ax2.set_ylabel('Accuracy (%)')
    ax2.legend()
    ax2.grid(True)
    
    # Learning rate schedule
    ax3.plot(epochs, history['learning_rates'], 'g-', label='Learning Rate')
    ax3.set_title('Learning Rate Schedule (Cosine Annealing)')
    ax3.set_xlabel('Epoch')
    ax3.set_ylabel('Learning Rate')
    ax3.legend()
    ax3.grid(True)
    
    # Model architecture visualization
    ax4.text(0.1, 0.9, 'CNN Architecture:', transform=ax4.transAxes, fontsize=12, fontweight='bold')
    architecture_text = """
Input: 1×28×28 (MNIST)
↓
Conv2d(1→32, 3×3) + BN + ReLU + MaxPool
↓ 32×14×14
Conv2d(32→64, 3×3) + BN + ReLU + MaxPool  
↓ 64×7×7
Conv2d(64→128, 3×3) + BN + ReLU
↓ 128×7×7
Flatten → FC(6272→512) + ReLU + Dropout
↓
FC(512→256) + ReLU + Dropout
↓
FC(256→10) → Output
    """
    ax4.text(0.1, 0.8, architecture_text, transform=ax4.transAxes, fontsize=10, 
             verticalalignment='top', fontfamily='monospace')
    ax4.set_xlim(0, 1)
    ax4.set_ylim(0, 1)
    ax4.axis('off')
    
    plt.tight_layout()
    plt.show()


def visualize_predictions(model, test_loader, num_samples=8):
    """Visualize model predictions on test samples"""
    model.eval()
    
    # Get a batch of test data
    data_iter = iter(test_loader)
    images, labels = next(data_iter)
    images, labels = images.to(device), labels.to(device)
    
    # Get predictions
    with torch.no_grad():
        outputs = model(images)
        _, predicted = torch.max(outputs, 1)
        probabilities = F.softmax(outputs, dim=1)
    
    # Plot samples
    fig, axes = plt.subplots(2, 4, figsize=(12, 6))
    axes = axes.ravel()
    
    for i in range(num_samples):
        # Denormalize image for display
        img = images[i].cpu().squeeze()
        img = img * 0.3081 + 0.1307  # Reverse normalization
        
        axes[i].imshow(img, cmap='gray')
        axes[i].set_title(f'True: {labels[i].item()}, Pred: {predicted[i].item()}\n'
                         f'Confidence: {probabilities[i][predicted[i]].item():.3f}')
        axes[i].axis('off')
    
    plt.tight_layout()
    plt.show()


def main():
    """Main training script"""
    # Hyperparameters (all initialized parameters from theory)
    BATCH_SIZE = 128
    NUM_EPOCHS = 20
    INITIAL_LR = 0.001
    DROPOUT_PROB = 0.5
    
    print("=" * 60)
    print("COMPLETE CNN TRAINING ON MNIST")
    print("=" * 60)
    print(f"Batch Size: {BATCH_SIZE}")
    print(f"Number of Epochs: {NUM_EPOCHS}")
    print(f"Initial Learning Rate: {INITIAL_LR}")
    print(f"Dropout Probability: {DROPOUT_PROB}")
    print(f"Device: {device}")
    print("=" * 60)
    
    # Load data
    print("Loading MNIST dataset...")
    train_loader, test_loader = load_data(BATCH_SIZE)
    print(f"Training samples: {len(train_loader.dataset)}")
    print(f"Test samples: {len(test_loader.dataset)}")
    
    # Initialize model
    print("\nInitializing CNN model...")
    model = CNN(num_classes=10, dropout_prob=DROPOUT_PROB).to(device)
    
    # Print model summary
    total_params = sum(p.numel() for p in model.parameters())
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"Total parameters: {total_params:,}")
    print(f"Trainable parameters: {trainable_params:,}")
    
    # Train model
    print("\nStarting training...")
    start_time = time.time()
    history = train_model(model, train_loader, test_loader, NUM_EPOCHS, INITIAL_LR)
    training_time = time.time() - start_time
    
    print(f"\nTraining completed in {training_time:.2f} seconds")
    print(f"Final test accuracy: {history['test_accuracies'][-1]:.2f}%")
    
    # Plot results
    print("\nPlotting training history...")
    plot_training_history(history)
    
    # Visualize predictions
    print("Visualizing predictions...")
    visualize_predictions(model, test_loader)
    
    # Save model
    model_path = 'mnist_cnn_model.pth'
    torch.save({
        'model_state_dict': model.state_dict(),
        'history': history,
        'model_config': {
            'num_classes': 10,
            'dropout_prob': DROPOUT_PROB
        }
    }, model_path)
    print(f"\nModel saved to {model_path}")
    
    return model, history


if __name__ == "__main__":
    # Run the complete training pipeline
    model, history = main()
    
    print("\n" + "=" * 60)
    print("TRAINING COMPLETED SUCCESSFULLY!")
    print("=" * 60)
    print(f"Best test accuracy: {max(history['test_accuracies']):.2f}%")
    print("Model ready for inference!")
```
